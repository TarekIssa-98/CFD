
# Analytical Solution for Heat Conduction with Temperature-Dependent Conductivity

## Problem Statement

Consider a plane wall with temperature-dependent thermal conductivity and uniform internal heat generation. The governing equation is:

$$ \frac{\partial}{\partial x} \left[ k(T) \frac{\partial T}{\partial x} \right] + q'' = 0, \quad 0 \leq x \leq L $$

with boundary conditions $T(0) = T_1$ and $T(L) = T_2$.


## Given Data

- $L = 0.1 \, \text{m}$
- $k_0 = 15 \, \text{W/(m·K)}$
- $\beta = 0.002 \, \text{K}^{-1}$
- $q'' = 2 \times 10^5 \, \text{W/m}^3$
- $T_1 = 250^\circ \text{C}$
- $T_2 = 50^\circ \text{C}$

## Analytical Solution Using Kirchhoff Transform

The temperature-dependent conductivity is given by:

$$ k(T) = k_0(1 + \beta T) $$

Introduce the Kirchhoff transform:

$$ \Theta(T) = \int_0^T \frac{k(T')}{k_0} dT' = T + \frac{\beta}{2} T^2 $$

This transformation linearizes the governing equation:

$$ \frac{d^2\Theta}{dx^2} = -\frac{q''}{k_0} $$

Integrating twice and applying boundary conditions gives:

$$ \Theta(x) = \Theta_1 + (\Theta_2 - \Theta_1) \frac{x}{L} + \frac{q''}{2k_0} x(L-x) $$

where:
- $\Theta_1 = T_1 + \frac{\beta}{2} T_1^2$
- $\Theta_2 = T_2 + \frac{\beta}{2} T_2^2$

Finally, inverting the Kirchhoff transform yields the exact temperature distribution:

$$ T(x) = \frac{-1 + \sqrt{1 + 2\beta\Theta(x)}}{\beta} $$

## Verification

For $\beta = 0$, the solution reduces to the familiar constant-property result:

$$ T(x) = T_1 + (T_2 - T_1) \frac{x}{L} + \frac{q''}{2k_0} x(L-x) $$

## Part (a): Derivation of Kirchhoff Transform Solution

### Step 1: Governing Equation:

$$ \frac{\partial}{\partial x} \left[ k_0 (1 + \beta T) \frac{\partial T}{\partial x} \right] + q'' = 0, \quad 0 \leq x \leq L $$

We have:

$$ (1 + \beta T) \frac{dT}{dx} = \frac{dT}{dx} + \beta T \frac{dT}{dx} $$

But note that:

$$ \frac{d}{dx} \left( T + \frac{\beta T^2}{2} \right) = \frac{dT}{dx} + \beta T \frac{dT}{dx} $$

Therefore:

$$ (1 + \beta T) \frac{dT}{dx} = \frac{d}{dx} \left( T + \frac{\beta T^2}{2} \right) $$

### Step 2: Apply to governing equation

$$ k_0 \frac{\partial}{\partial x} \left[ \frac{\partial}{\partial x} \left( T + \frac{\beta T^2}{2} \right) \right] + q'' = 0 $$

$$ k_0 \frac{d^2}{dx^2} \left( T + \frac{\beta T^2}{2} \right) + q'' = 0 $$

$$ \frac{d^2}{dx^2} \left( T + \frac{\beta T^2}{2} \right) + \frac{q''}{k_0} = 0 $$

### Step 3: Define Kirchhoff Parameter

Define:

$$ \Theta = T + \frac{\beta T^2}{2} $$

Therefore:

$$ \frac{d^2 \Theta}{dx^2} = -\frac{q''}{k_0} $$

### Step 4: Boundary conditions in terms of $\Theta$

At $x = 0$:

$$ \Theta(0) = \Theta_1 = T_1 + \frac{\beta T_1^2}{2} $$

At $x = L$:

$$ \Theta(L) = \Theta_2 = T_2 + \frac{\beta T_2^2}{2} $$

### Step 5: Integrate twice

First integration:

$$ \frac{d\Theta}{dx} = -\frac{q''}{k_0} x + C_1 $$

Second integration:

$$ \Theta(x) = -\frac{q''}{2k_0} x^2 + C_1 x + C_2 $$

### Step 6: Apply boundary conditions to find constants

At $x = 0 \implies \Theta(0) = \Theta_1$:

$$ \Theta_1 = -\frac{q''}{2k_0}(0)^2 + C_1(0) + C_2 $$

$$ \Theta_1 = 0 + 0 + C_2 \implies C_2 = \Theta_1 $$

At $x = L \implies \Theta(L) = \Theta_2$:

$$ \Theta_2 = -\frac{q''L^2}{2k_0} + C_1 L + \Theta_1 $$

$$ C_1 L = \Theta_2 - \Theta_1 + \frac{q''L^2}{2k_0} $$

$$ C_1 = \frac{\Theta_2 - \Theta_1}{L} + \frac{q''L}{2k_0} $$

### Step 7: Substitute constants back into $\Theta(x)$

Substituting $C_1$ and $C_2$ into $\Theta(x)$:

$$ \Theta(x) = -\frac{q''}{2k_0} x^2 + \left[ \frac{\Theta_2 - \Theta_1}{L} + \frac{q''L}{2k_0} \right] x + \Theta_1 $$

$$ \Theta(x) = -\frac{q''}{2k_0} x^2 + \frac{q''L}{2k_0} x + \Theta_1 + (\Theta_2 - \Theta_1)\frac{x}{L} $$

$$ \Theta(x) = \frac{q''}{2k_0} \left( xL - x^2 \right) + \Theta_1 + (\Theta_2 - \Theta_1)\frac{x}{L} $$

$$ \boxed{\Theta(x) = \frac{q''}{2k_0} x(L - x) + \Theta_1 + (\Theta_2 - \Theta_1)\frac{x}{L}} $$

### Step 8: Invert to find $T(x)$

We know:

$$ \Theta = T + \frac{\beta}{2} T^2 $$

Rearranging:

$$ \frac{\beta}{2} T^2 + T - \Theta = 0 $$

This is a quadratic equation in $T$:

$$ aT^2 + bT + c = 0 $$

where:

$$ a = \frac{\beta}{2}, \quad b = 1, \quad c = -\Theta $$

Using the quadratic formula:

$$ T = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a} $$

$$ T = \frac{-1 \pm \sqrt{1 - 4\left(\frac{\beta}{2}\right)(-\Theta)}}{2\left(\frac{\beta}{2}\right)} $$

$$ T = \frac{-1 \pm \sqrt{1 + 2\beta\Theta}}{\beta} $$

Since $T$ must be positive for a physically valid situation, we take the positive root:

$$ \boxed{T(x) = \frac{-1 + \sqrt{1 + 2\beta\Theta(x)}}{\beta}} $$

## Part (b): Discretization of the Nonlinear Equation

### Governing Equation:

$$ \frac{d}{dx} \left[ k(T) \frac{dT}{dx} \right] + q'' = 0 $$

## Step 1: Grid Definition

Divide the domain $0 \leq x \leq L$ into $N$ equal intervals:

$$ \Delta x = \frac{L}{N} $$

Nodes are: $0, 1, 2, 3, ..., N$

Where:
- Node $0$ is at $x = 0$ (left boundary, known temperature $T_1$)
- Node $N$ is at $x = L$ (right boundary, known temperature $T_2$)
- Interior nodes: $i = 1, 2, 3, ..., N-1$

---

## Step 2: Discretization of the Derivative Term

Using finite difference method:

$$ \frac{d}{dx} \left[ k(T) \frac{dT}{dx} \right]_i = \frac{\left( k \frac{dT}{dx} \right)_{i+\frac{1}{2}} - \left( k \frac{dT}{dx} \right)_{i-\frac{1}{2}}}{\Delta x} $$

### Temperature Gradients:

$$ \left. \frac{dT}{dx} \right|_{i+\frac{1}{2}} = \frac{T_{i+1} - T_i}{\Delta x} $$

$$ \left. \frac{dT}{dx} \right|_{i-\frac{1}{2}} = \frac{T_i - T_{i-1}}{\Delta x} $$

---

## Step 3: Substitute into the Governing Equation

$$ \frac{k_{i+\frac{1}{2}} \left( \frac{T_{i+1} - T_i}{\Delta x} \right) - k_{i-\frac{1}{2}} \left( \frac{T_i - T_{i-1}}{\Delta x} \right)}{\Delta x} + q'' = 0 $$

$$ k_{i+\frac{1}{2}} (T_{i+1} - T_i) - k_{i-\frac{1}{2}} (T_i - T_{i-1}) + q'' (\Delta x)^2 = 0 $$

---

## Step 4: Final Discretized Equation

$$ \boxed{ -k_{i-\frac{1}{2}} T_{i-1} + (k_{i+\frac{1}{2}} + k_{i-\frac{1}{2}}) T_i - k_{i+\frac{1}{2}} T_{i+1} = q'' (\Delta x)^2 } $$

---

## Step 5: Thermal Conductivity at Interfaces

### At the right interface $(i + \frac{1}{2})$:

$$ \boxed{ k_{i+\frac{1}{2}} = k_0 \left[ 1 + \beta \left( \frac{T_{i+1} + T_i}{2} \right) \right] } $$

### At the left interface $(i - \frac{1}{2})$:

$$ \boxed{ k_{i-\frac{1}{2}} = k_0 \left[ 1 + \beta \left( \frac{T_i + T_{i-1}}{2} \right) \right] } $$

---

## Step 6: Apply Boundary Conditions

### At the left boundary $(i = 0)$:

$$ T_0 = T_1  $$

### At the right boundary $(i = N)$:

$$ T_N = T_2  $$

---

## Step 7: Matrix Form for $N = 5$ (5 nodes total: $T_0, T_1, T_2, T_3, T_4$)

**Unknowns:** $T_1, T_2, T_3$ (interior nodes)
**Knowns:** $T_0 = T_1$ (given), $T_4 = T_2$ (given)

### Interface Conductivities:

$$ k_{1/2} = k_0 \left[ 1 + \beta \left( \frac{T_0 + T_1}{2} \right) \right] $$

$$ k_{3/2} = k_0 \left[ 1 + \beta \left( \frac{T_1 + T_2}{2} \right) \right] $$

$$ k_{5/2} = k_0 \left[ 1 + \beta \left( \frac{T_2 + T_3}{2} \right) \right] $$

$$ k_{7/2} = k_0 \left[ 1 + \beta \left( \frac{T_3 + T_4}{2} \right) \right] $$

---

### Node 1 ($i = 1$):

$$ -k_{1/2} T_0 + (k_{3/2} + k_{1/2}) T_1 - k_{3/2} T_2 = q'' (\Delta x)^2 $$

$$ \boxed{ (k_{3/2} + k_{1/2}) T_1 - k_{3/2} T_2 = q'' (\Delta x)^2 + k_{1/2} T_0 } $$

---

### Node 2 ($i = 2$):

$$ -k_{3/2} T_1 + (k_{5/2} + k_{3/2}) T_2 - k_{5/2} T_3 = q'' (\Delta x)^2 $$

$$ \boxed{ -k_{3/2} T_1 + (k_{3/2} + k_{5/2}) T_2 - k_{5/2} T_3 = q'' (\Delta x)^2 } $$

---

### Node 3 ($i = 3$):

$$ -k_{5/2} T_2 + (k_{7/2} + k_{5/2}) T_3 - k_{7/2} T_4 = q'' (\Delta x)^2 $$

$$ \boxed{ -k_{5/2} T_2 + (k_{5/2} + k_{7/2}) T_3 = q'' (\Delta x)^2 + k_{7/2} T_4 } $$

---

## Step 8: Matrix Form $A \cdot T = b$

$$ \boxed{ \begin{bmatrix}
(k_{3/2} + k_{1/2}) & -k_{3/2} & 0 \\
-k_{3/2} & (k_{3/2} + k_{5/2}) & -k_{5/2} \\
0 & -k_{5/2} & (k_{5/2} + k_{7/2})
\end{bmatrix}
\begin{bmatrix}
T_1 \\
T_2 \\
T_3
\end{bmatrix}
=
\begin{bmatrix}
q'' (\Delta x)^2 + k_{1/2} T_0 \\
q'' (\Delta x)^2 \\
q'' (\Delta x)^2 + k_{7/2} T_4
\end{bmatrix} } $$

---

## Step 9: Picard Iteration Method

Using **conductivities at iteration level P** and **temperatures at iteration level P+1**:
The initial guess should be the linear profile between the two boundary temperatures:
 $$T_{(x)}^{(0)} = T_1 + (T_2 - T_1)  \frac{x}{L}$$


### Step 9a: Compute Conductivities at Level P

$$ k_{1/2}^{(P)} = k_0 \left[ 1 + \beta \left( \frac{T_0^{(P)} + T_1^{(P)}}{2} \right) \right] $$

$$ k_{3/2}^{(P)} = k_0 \left[ 1 + \beta \left( \frac{T_1^{(P)} + T_2^{(P)}}{2} \right) \right] $$

$$ k_{5/2}^{(P)} = k_0 \left[ 1 + \beta \left( \frac{T_2^{(P)} + T_3^{(P)}}{2} \right) \right] $$

$$ k_{7/2}^{(P)} = k_0 \left[ 1 + \beta \left( \frac{T_3^{(P)} + T_4^{(P)}}{2} \right) \right] $$

### Step 9b: Solve for Temperatures at Level P+1

$$ \begin{bmatrix}
(k_{3/2}^{(P)} + k_{1/2}^{(P)}) & -k_{3/2}^{(P)} & 0 \\
-k_{3/2}^{(P)} & (k_{3/2}^{(P)} + k_{5/2}^{(P)}) & -k_{5/2}^{(P)} \\
0 & -k_{5/2}^{(P)} & (k_{5/2}^{(P)} + k_{7/2}^{(P)})
\end{bmatrix}
\begin{bmatrix}
T_1^{(P+1)} \\
T_2^{(P+1)} \\
T_3^{(P+1)}
\end{bmatrix}
=
\begin{bmatrix}
q'' (\Delta x)^2 + k_{1/2}^{(P)} T_0^{(P)} \\
q'' (\Delta x)^2 \\
q'' (\Delta x)^2 + k_{7/2}^{(P)} T_4^{(P)}
\end{bmatrix} $$

### Step 9c: Check Convergence

$$ \max_i |T_i^{(P+1)} - T_i^{(P)}| < \epsilon $$

If not converged, set $P = P+1$ and repeat steps.

## Equations Summary

### Discretized Equation:

$$ \boxed{ -k_{i-\frac{1}{2}} T_{i-1} + (k_{i+\frac{1}{2}} + k_{i-\frac{1}{2}}) T_i - k_{i+\frac{1}{2}} T_{i+1} = q'' (\Delta x)^2 } $$

### Interface Conductivities:

$$ \boxed{ k_{i+\frac{1}{2}} = k_0 \left[ 1 + \beta \left( \frac{T_{i+1} + T_i}{2} \right) \right] } $$

$$ \boxed{ k_{i-\frac{1}{2}} = k_0 \left[ 1 + \beta \left( \frac{T_i + T_{i-1}}{2} \right) \right] } $$

### Matrix Form (General):

$$ {\small
\begin{bmatrix}
(k_{3/2} + k_{1/2}) & -k_{3/2} & 0 & \cdots & 0 \\
-k_{3/2} & (k_{3/2} + k_{5/2}) & -k_{5/2} & \cdots & 0 \\
0 & -k_{5/2} & (k_{5/2} + k_{7/2}) & \cdots & 0 \\
\vdots & \vdots & \vdots & \ddots & \vdots \\
0 & 0 & 0 & \cdots & (k_{N-1/2} + k_{N-3/2})
\end{bmatrix}
\begin{bmatrix}
T_1 \\
T_2 \\
T_3 \\
\vdots \\
T_{N-1}
\end{bmatrix}
= \begin{bmatrix}
q'' (\Delta x)^2 + k_{1/2} T_0 \\
q'' (\Delta x)^2 \\
q'' (\Delta x)^2 \\
\vdots \\
q'' (\Delta x)^2 + k_{N-1/2} T_N
\end{bmatrix}
} $$
### Picard Iteration:

$$ \boxed{ A(T^{(P)}) \cdot T^{(P+1)} = b(T^{(P)}) } $$

where $A$ and $b$ are evaluated using temperatures from iteration level $P$, and we solve for temperatures at iteration level $P+1$.

```python
# -*- coding: utf-8 -*-
"""
Created on Thu Sep  3 23:07:51 2026

@author: tarek
"""

import numpy as np
import matplotlib.pyplot as plt
import os

# Physical / problem constants (SI units unless noted)
L = 0.1                      # m,  wall thickness
k0 = 15.0                    # W/(m*K), reference conductivity at T = 0
q_gen = 2.0e5                # W/m^3, uniform volumetric heat generation
T1_K = 523.15                # K (250 C), left boundary temperature
T2_K = 323.15                # K (50 C),  right boundary temperature
```

```python
# TASK b: EXACT KIRCHHOFF SOLUTION 


def exact_kirchhoff_solution(beta, T1, T2, x):
    """
    Exact solution using the Kirchhoff transform Theta = T + (beta/2) T^2,
    derived in Part (a). Used only to *check* the numerical scheme in
    Part (b)-(d)

    Parameters
    ----------
    beta : float, temperature coefficient of conductivity (K^-1)
    T1, T2 : float, boundary temperatures (K)
    x : array, spatial coordinates (m)

    Returns
    -------
    T : temperature field (K)
    theta : Kirchhoff transform field
    """
    if beta == 0:
        # beta = 0 collapses to the familiar constant-property parabola
        T = T1 + (T2 - T1) * x / L + q_gen / (2 * k0) * x * (L - x)
        return T, T

    theta1 = T1 + beta * T1**2 / 2
    theta2 = T2 + beta * T2**2 / 2
    theta = theta1 + (theta2 - theta1) * x / L + q_gen / (2 * k0) * x * (L - x)
    # invert the quadratic Theta = T + (beta/2)T^2 for the physical (positive) root
    T = (-1 + np.sqrt(1 + 2 * beta * theta)) / beta
    return T, theta
```

```python
# ============================================================================
# TASK b: THOMAS ALGORITHM (tridiagonal direct solver)
# ============================================================================

def thomas_algorithm(a, b, c, d):
    """
    Thomas algorithm (tridiagonal Gaussian elimination) for
        a[i]*T[i-1] + b[i]*T[i] + c[i]*T[i+1] = d[i],   a[0] = c[n-1] = 0

    This is a *direct* solver: O(n) operations, no iteration needed, and
    exact to machine precision. It solves the *linear* system produced by one Picard step,
    i.e. A(T^p) T^(p+1) = b(T^p), with face conductivities is at the
    previous Picard level p.
    """
    n = len(d)
    c_prime = np.zeros(n - 1)
    d_prime = np.zeros(n)

    c_prime[0] = c[0] / b[0]
    d_prime[0] = d[0] / b[0]

    for i in range(1, n):
        denom = b[i] - a[i] * c_prime[i - 1]
        if i < n - 1:
            c_prime[i] = c[i] / denom
        d_prime[i] = (d[i] - a[i] * d_prime[i - 1]) / denom

    T = np.zeros(n)
    T[-1] = d_prime[-1]
    for i in range(n - 2, -1, -1):
        T[i] = d_prime[i] - c_prime[i] * T[i + 1]
    return T


def build_tridiagonal_system(T, beta, N, dx):
    """
    Assemble the tridiagonal system A(T) T_new = b(T) for the interior
    nodes i = 1..N-1, using face conductivities k_{i+-1/2} evaluated from
    the *current* temperature guess T (this "freezing" is exactly the
    Picard linearization). Returns the diagonals (a,b,c), RHS d, and the
    face-conductivity array k_half (k_half[i] = k_{i-1/2}) for reuse by
    the energy-balance check and the Gauss-Seidel cross-check below.
    """
    k_half = np.zeros(N + 1)
    for i in range(N):
        k_half[i + 1] = k0 * (1 + beta * (T[i] + T[i + 1]) / 2)

    n_inner = N - 1
    a = np.zeros(n_inner)
    b = np.zeros(n_inner)
    c = np.zeros(n_inner)
    d = np.zeros(n_inner)

    for i in range(1, N):
        idx = i - 1
        k_left = k_half[i]       # k_{i-1/2}
        k_right = k_half[i + 1]  # k_{i+1/2}

        if i == 1:
            # first interior node: T_{i-1} = T_0 is a *known* boundary value,
            # so its contribution -k_left*T_0 moves to the RHS
            b[idx] = k_left + k_right
            c[idx] = -k_right
            d[idx] = q_gen * dx**2 + k_left * T[0]
        elif i == N - 1:
            # last interior node: T_{i+1} = T_N is the known right boundary value
            a[idx] = -k_left
            b[idx] = k_left + k_right
            d[idx] = q_gen * dx**2 + k_right * T[N]
        else:
            # a genuine interior node: standard tridiagonal row
            a[idx] = -k_left
            b[idx] = k_left + k_right
            c[idx] = -k_right
            d[idx] = q_gen * dx**2

    return a, b, c, d, k_half



#  PICARD SOLVER

def picard_solver(beta, N=20, tol=1e-6, max_iter=100):
    """
    Picard (successive-substitution) iteration for the nonlinear
    conduction problem d/dx[k(T) dT/dx] + q_gen = 0, discretized directly
    with k evaluated at cell faces:
        k_{i+1/2} = k0 * [1 + beta*(T_i + T_{i+1})/2]

    Outer loop (this function): update k_{i+-1/2} from the latest T guess,
    then solve the resulting *linear* tridiagonal system exactly with the
    Thomas algorithm, and repeat until max|T^(p+1) - T^p| < tol.

    Returns
    -------
    T : converged temperature field (K)
    x : node coordinates (m)
    iterations : number of Picard iterations actually used
    errors : max|dT| recorded at every iteration (for convergence plots)
    k_half : face-conductivity array from the final iteration
    """
    dx = L / N
    x = np.linspace(0, L, N + 1)

    # Initial guess: the constant-property (linear) profile -- a physically
    # reasonable starting point that is exact when beta = 0
    T = T1_K + (T2_K - T1_K) * x / L

    errors = []
    k_half = None

    for iteration in range(max_iter):
        T_old = T.copy()

        a, b, c, d, k_half = build_tridiagonal_system(T, beta, N, dx)
        T_inner = thomas_algorithm(a, b, c, d)
        T[1:N] = T_inner

        error = np.max(np.abs(T - T_old))
        errors.append(error)

        if error < tol:
            return T, x, iteration + 1, errors, k_half

    return T, x, max_iter, errors, k_half
```

## Discussion on the results

**1. Effect on temperature curvature**
- $\beta = 0$: parabolic temperature profile corresponding to constant thermal conductivity..
- $\beta > 0$ (conductivity increases with temperature): the hotter regions conduct heat more effectively, resulting in a slightly flatter/less curved temperature profile and a lower temperature near the middle compared with $\beta$ = 0.
- $\beta < 0$ (conductivity decreases with temperature): the hotter regions conduct heat less effectively, causing greater curvature and a higher temperature near the middle compared with $\beta$ = 0.

**2. Convergence behaviour (actual counts,  summary table
above `iteration_summary`, N = 20, tol = $10^{-6}$)**

| $\beta$ (K$^{-1}$) | Picard iterations |
|---|---|
| 0        | 2 |
| 0.002    | 6 |
| 0.005    | 7 |
| -0.001   | 7 |

$\beta = 0$ needs 2 passes rather than "1", the problem is linear so the
*first* Thomas solve already gives the exact answer, but a second pass is
still required to confirm max$|T^{(p+1)}-T^{(p)}|<$tol against the initial
linear guess. For $\beta\neq0$, more Picard sweeps are needed because each
sweep only *partially* updates the face conductivities toward their
self-consistent values.

The trend is not simply "more iterations for larger
$|\beta|$: $\beta=-0.001$ (smaller magnitude) takes as many iterations as
$\beta=0.005$ (much larger magnitude). What actually controls the Picard
convergence rate is how strongly $k(T)$ responds to the *change* in T
between sweeps near the current solution -- both the sign and the local
temperature range matter, not just $|\beta|$ in isolation.

**3. Physical interpretation**
- Positive $\beta$: a local T rise
  raises $k$, which conducts that excess heat away and pulls T back down.
- Negative $\beta$:a local T rise
  lowers $k$, trapping heat and pushing T up further, which is also why the
  numerics need to work slightly harder to pin down the self-consistent
  field.

**4. The numerical scheme matches the exact solution (Plot 2), regardless of grid size**

Expanding the face-averaged conductivity term:

Starting with the face-averaged thermal conductivity at the interface between nodes $i$ and $i+1$:

$$k_{i+\frac{1}{2}} = k_0\left(1 + \frac{\beta}{2}(T_i + T_{i+1})\right)$$

Now, expand the flux term $k_{i+\frac{1}{2}}(T_{i+1} - T_i)$:

$$k_{i+\frac{1}{2}}(T_{i+1} - T_i) = k_0(T_{i+1} - T_i) + \frac{k_0\beta}{2}(T_i + T_{i+1})(T_{i+1} - T_i)$$

Using the identity $(T_i + T_{i+1})(T_{i+1} - T_i) = T_{i+1}^2 - T_i^2$:

$$k_{i+\frac{1}{2}}(T_{i+1} - T_i) = k_0(T_{i+1} - T_i) + \frac{k_0\beta}{2}(T_{i+1}^2 - T_i^2)$$

Factor out $k_0$:

$$k_{i+\frac{1}{2}}(T_{i+1} - T_i) = k_0\left[(T_{i+1} - T_i) + \frac{\beta}{2}(T_{i+1}^2 - T_i^2)\right]$$

$$k_{i+\frac{1}{2}}(T_{i+1} - T_i) = k_0\left[\left(T_{i+1} + \frac{\beta}{2}T_{i+1}^2\right) - \left(T_i + \frac{\beta}{2}T_i^2\right)\right]$$

Define the transformation:

$$\Theta(T) = T + \frac{\beta}{2}T^2$$

Therefore:

$$k_{i+\frac{1}{2}}(T_{i+1} - T_i) = k_0(\Theta_{i+1} - \Theta_i)$$

This shows that the nonlinear flux term simplifies to a linear flux in the transformed variable.

so the finite-difference equation in Task (b) is, term for term, *exactly*
the standard 3-point central difference of $d^2\Theta/dx^2=-q''/k_0$ from
Part (a). Since $\Theta(x)$ is quadratic in $x$, a central difference is
exact for it on any grid -- so this particular discretization has **zero
discretization error** here, independent of $N$. That is a special feature
of $k(T)$ being linear in $T$ with a spatially uniform $q''$, not a general
property of the scheme; with a more general $k(T)$ this exactness would not
hold and the usual $O(\Delta x^2)$ convergence would be visible instead.
